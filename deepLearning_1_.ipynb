The following code demonstrates the basics of a feedforward neural network using nn.Sequential. The model (commonly named model) is built using modules from the nn package. To perform a forward pass, we pass the input tensor directly to the model and obtain the output. Input tensors should typically be floating-point types, most commonly float32, when used with neural network layers.

```python
import torch
import torch.nn as nn

# Create an input tensor
input_tensor = torch.tensor(
    [[2, 3, 6, 7, 9, 3, 2, 1]],
    dtype=torch.float32
)

print("Input shape:", input_tensor.shape)
print("Input dtype:", input_tensor.dtype)

# Define a simple neural network
model = nn.Sequential(
    nn.Linear(8, 4),   # Hidden layer
    nn.ReLU(),         # Activation function
    nn.Linear(4, 1)    # Output layer
)

# Forward pass
output = model(input_tensor)

print("\nOutput:", output)
print("Output shape:", output.shape)
print("Output dtype:", output.dtype)
```

The .numel() method returns the number of parameters, so for calculating the tottal number of parameters in the network we shall:

```python
# Count the total number of parameters in the network
total_params = sum(param.numel() for param in model.parameters())

print("Total number of parameters:", total_params)
```
